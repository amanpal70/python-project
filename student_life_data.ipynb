BIVARIATE ANALYSIS OF **VARIABLES**

```python
student_df.groupby('Stress_Level')['Sleep_Hours_Per_Day'].mean().plot(kind='bar')

plt.title('Average Sleep Hours vs Stress Level')
plt.xlabel('Stress Level')
plt.ylabel('Average Sleep Hours Per Day')

plt.show()
```

agar ham 7 hours se kam sleep krte h to stress level high h

```python
sns.scatterplot( x='Study_Hours_Per_Day', y='GPA',data=student_df)
plt.title('Study Hours vs GPA')
plt.xlabel('Study Hours Per Day')
plt.ylabel('GPA')
plt.show()
```

2 numerical vlaues ko compare krne k liye scatter plot use kiya h jisse ki hame pta chalta h ki ki jitne jyada study hours honge utne gpa achi hogi jo ki top right corner

```python
sns.scatterplot(
    x='Physical_Activity_Hours_Per_Day',
    y='Study_Hours_Per_Day',
    data=student_df
)

plt.title('Physical Activity Hours vs Study Hours')
plt.xlabel('Physical Activity Hours Per Day')
plt.ylabel('Study Hours Per Day')
plt.show()
```

less number of physical activity hours more the number of study hours

MULTIVARIATE

```python
sns.scatterplot(
    x='Study_Hours_Per_Day',
    y='GPA',
    hue='Physical_Activity_Hours_Per_Day',
    data=student_df
)

plt.title('Study Hours, GPA and Physical Activity')
plt.xlabel('Study Hours Per Day')
plt.ylabel('GPA')
plt.show()
```

this plot shows that how physical activity hours and study hours per day are related to gpa of student

```python
correlation = student_df[
    ['Study_Hours_Per_Day',
     'Extracurricular_Hours_Per_Day',
     'Sleep_Hours_Per_Day',
     'Social_Hours_Per_Day',
     'Physical_Activity_Hours_Per_Day',
     'GPA']
].corr()

plt.figure(figsize=(8,6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap')
plt.show()
```

This is a data analysis of lifestyle of student in which various columns , index which shows the relationship between each other and how they effect .If giving major hours of a day to one  activity makes an impact on on gpa stress level and study hours and so on.
